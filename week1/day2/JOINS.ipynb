```sql
CREATE TABLE likki_employees (
emp_id INT PRIMARY KEY,
emp_name VARCHAR(50),
manager_id INT,
dept_id INT
);


```

```sql
select e.emp_name,m.manager_id from likki_employees e  join likki_employees m on e.emp_id=m.emp_id
```

```sql
CREATE TABLE departments (
dept_id INT PRIMARY KEY,
dept_name VARCHAR(50)
);
```

```sql
select e.emp_name,d.dept_name from likki_employees e left join departments d on e.dept_id=d.dept_id
```

```sql
select e.emp_name,m.manager_id from likki_employees e  join likki_employees m on e.manager_id=m.manager_id
```

```sql
select e.emp_name,d.dept_name from likki_employees e left join departments d on e.dept_id=d.dept_id where d.dept_id=null
```

```sql
CREATE TABLE projects (
project_id INT PRIMARY KEY,
project_name VARCHAR(50),
emp_id INT
);

```

```sql
INSERT INTO projects (project_id, project_name, emp_id) VALUES
(1, 'Project A', 1),
(2, 'Project B', 2),
(3, 'Project C', 3),
(4, 'Project D', 4),
(5, 'Project E', 5);
```

```sql
select e.emp_name,p.project_name from likki_employees e  left join projects p on e.emp_id=p.emp_id
```

```sql
select e.emp_name,p.project_name from likki_employees e  right join projects p on e.emp_id=p.emp_id
```

```sql
select e.emp_name,p.project_name from likki_employees e  left join projects p on e.emp_id=p.emp_id  where p.project_id is null
```

```sql
select e.emp_name e.manager_id from likki_employees e  right join projects p on e.emp_id=p.emp_id
where e.manager_id is not null
```

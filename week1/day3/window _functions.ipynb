```sql
--1
SELECT emp_name,salary,
ROW_NUMBER()
  OVER(ORDER by salary DESC) as ROW_NUMBER
FROM employees_2
```

```sql
SELECT emp_name,department,
ROW_NUMBER()
  OVER(PARTITION by department ORDER by salary DESC) as ROW_NUMBER
FROM employees_2
```

```sql
SELECT emp_name,join_date,
ROW_NUMBER()
  OVER( ORDER by join_date DESC) as ROW_NUMBER
FROM employees_2
```

```sql
SELECT emp_name, department,join_date,
ROW_NUMBER()
  OVER( PARTITION by department ORDER by join_date ASC) as ROW_NUMBER
FROM employees_2
```

### RANK()

```sql
SELECT emp_name,salary,
RANK()
  OVER(ORDER by salary DESC) as RANK
FROM employees_2
```

```sql
SELECT emp_name,department,
RANK()
  OVER(PARTITION by department ORDER by salary DESC) as RANK
FROM employees_2
```

```sql
SELECT emp_name,join_date,
RANK()
  OVER( ORDER by join_date DESC) as RANK
FROM employees_2
```

```sql
SELECT emp_name, department,salary,
RANK()
  OVER( PARTITION by department ORDER by salary ASC) as RANK
FROM employees_2
```

```sql
SELECT emp_name, 
RANK()
  OVER( ORDER by emp_name  ASC) as RANK
FROM employees_2
```

### DENSE_RANK()
